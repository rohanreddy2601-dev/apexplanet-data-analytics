```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')
```

```python
categorical_cols = ['Ship Mode', 'Segment', 'Market', 'Region', 'Category', 'Sub-Category', 'Order Priority']
df[categorical_cols] = df[categorical_cols].astype('category')
```

```python
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return data[(data[column] >= lower) & (data[column] <= upper)]

print("Rows before outlier removal:", df.shape[0])
df_clean = remove_outliers_iqr(df, 'Sales')
df_clean = remove_outliers_iqr(df_clean, 'Profit')
print("Rows after outlier removal:", df_clean.shape[0])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
df['Sales'].hist(bins=50)
plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(12,5))
df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum().plot()
plt.title('Monthly Sales Trend')
plt.ylabel('Total Sales')
plt.xlabel('Month')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[['Sales','Quantity','Discount','Profit','Shipping Cost']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(8,5))
df.groupby('Category', observed=True)['Sales'].sum().sort_values(ascending=False).plot(kind='bar')
plt.title('Total Sales by Category')
plt.ylabel('Total Sales')
plt.show()
```
