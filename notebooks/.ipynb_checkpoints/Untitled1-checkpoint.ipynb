```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv('data/ecomm_data_cleaned.csv')  # adjust path if needed
```

```python
monthly = df.groupby(pd.to_datetime(df['Order Date']).dt.to_period('M'))['Sales'].sum()
plt.figure(figsize=(12,5))
monthly.plot(kind='line')
plt.title('Monthly Sales Trend')
plt.ylabel('Sales')
plt.savefig('reports/monthly_sales_trend.png')
plt.show()
```

```python
df = pd.read_csv('../data/ecomm_data_cleaned.csv')
```

```python
plt.savefig('../reports/monthly_sales_trend.png')
```

```python
import os
os.makedirs('../reports', exist_ok=True)
```

```python
monthly = df.groupby(pd.to_datetime(df['Order Date']).dt.to_period('M'))['Sales'].sum()
plt.figure(figsize=(12,5))
monthly.plot(kind='line')
plt.title('Monthly Sales Trend')
plt.ylabel('Sales')
plt.savefig('../reports/monthly_sales_trend.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
df.groupby('Category')['Sales'].sum().sort_values(ascending=False).plot(kind='bar')
plt.title('Total Sales by Category')
plt.savefig('../reports/sales_by_category.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Sales'], df['Profit'], alpha=0.3)
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Sales vs Profit')
plt.savefig('../reports/sales_vs_profit.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
df['Sales'].hist(bins=50)
plt.title('Sales Distribution')
plt.savefig('../reports/sales_distribution.png')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[['Sales','Quantity','Discount','Profit','Shipping Cost']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.savefig('../reports/correlation_heatmap.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxenplot(x='Category', y='Profit', data=df)
plt.title('Profit Distribution by Category (Boxen Plot)')
plt.savefig('../reports/profit_boxen.png')
plt.show()
```

```python
sample = df[['Sales','Profit','Discount','Quantity']].sample(1000)
sns.pairplot(sample)
plt.savefig('../reports/pairplot.png')
plt.show()
```
